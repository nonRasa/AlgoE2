# تمرین دوم: نمودار و مقایسهٔ پیچیدگی زمانی

هدف این تمرین، دیدن تفاوت رشد الگوریتم‌هاست: اگر اندازهٔ ورودی بیشتر شود، تعداد گام‌های لازم چقدر تغییر می‌کند؟ ابتدا نمودارهای هشت مرتبهٔ رایج را مقایسه می‌کنیم؛ سپس برای هر مرتبه یک الگوریتم، مثال عددی، پیاده‌سازی و فلوچارت می‌بینیم.

## مفهوم پیچیدگی زمانی

پیچیدگی زمانی `Time Complexity` رشد تعداد عملیات را بر حسب اندازهٔ ورودی `Input Size` توصیف می‌کند. متغیر `n` در مثال‌های فهرستی تعداد اعضاست؛ در ضرب ماتریس، تعداد سطر یا ستون هر ماتریس را نشان می‌دهد. دسترسی اندیسی، مقایسه، جمع و ضرب عددی را در مدل شمارش این تمرین یک عملیات با هزینهٔ ثابت در نظر می‌گیریم.

نماد کران بالا `Big O` می‌گوید برای ورودی‌های به‌اندازهٔ کافی بزرگ، رشد زمان از یک تابع، تا ضریب ثابت، بیشتر نمی‌شود. نماد تتا `Theta` مرتبهٔ رشد دقیق را با کران بالا و پایین نشان می‌دهد. مثلاً تابع زیر هم کران بالای درجهٔ دو دارد و هم مرتبهٔ دقیق درجهٔ دو؛ جملهٔ درجهٔ دو با بزرگ‌شدن ورودی غالب می‌شود.

```text
T(n) = 3n² + 5n + 2
T(n) ∈ O(n²)
T(n) ∈ Θ(n²)
```

قرارداد مقایسه: جدول‌ها از نماد `Θ` برای مرتبهٔ دقیق استفاده می‌کنند و تحلیل مثال‌ها بر اساس بدترین حالت `Worst Case` است. پایهٔ لگاریتم در نمودارها دو است؛ تغییر پایهٔ ثابت تنها یک ضریب ثابت به تابع اضافه می‌کند.

## ترتیب از سریع‌ترین به کندترین

ترتیب زیر برای **نرخ رشد در ورودی‌های به‌اندازهٔ کافی بزرگ** است. هرچه تابع آهسته‌تر رشد کند، افزایش ورودی کار کمتری به الگوریتم تحمیل می‌کند.

```text
Θ(1) < Θ(log n) < Θ(n) < Θ(n log n) < Θ(n²) < Θ(n³) < Θ(2ⁿ) < Θ(n!)
```

توضیح رتبه‌ها: ثابت، لگاریتمی، خطی، خطی‌لگاریتمی، درجهٔ دو، درجهٔ سه، نمایی و فاکتوریلی. زمان واقعی یک اجرا به ضرایب ثابت، نوع عملیات و پیاده‌سازی هم وابسته است؛ از یکسان‌بودن مرتبهٔ دو برنامه، زمان یکسان نتیجه نمی‌شود.

نکتهٔ اندازه‌های کوچک: همین ترتیب ممکن است در چند مقدار کوچک برقرار نباشد. برای نمونه، در اندازهٔ چهار، تابع فاکتوریلی از تابع درجهٔ سه کوچک‌تر است؛ با بزرگ‌ترشدن ورودی، فاکتوریلی از آن پیشی می‌گیرد. مقایسهٔ مرتبه‌ها به رفتار رشد توجه دارد.

```text
n = 4:  n! = 24 < n³ = 64
n = 8:  n! = 40320 > n³ = 512
```

| رتبهٔ رشد | نام مرتبه | مرتبهٔ زمانی | مثال الگوریتم | معنای اندازه |
|---|---|---|---|---|
| جایگاه 1 | ثابت | `Θ(1)` | دسترسی مستقیم به عضو اول | تعداد اعضای فهرست |
| جایگاه 2 | لگاریتمی | `Θ(log n)` | جست‌وجوی دودویی | تعداد اعضای فهرست مرتب |
| جایگاه 3 | خطی | `Θ(n)` | یافتن بیشینه با پیمایش | تعداد اعضای فهرست |
| جایگاه 4 | خطی‌لگاریتمی | `Θ(n log n)` | مرتب‌سازی ادغامی | تعداد اعضای فهرست |
| جایگاه 5 | درجهٔ دو | `Θ(n²)` | مرتب‌سازی انتخابی | تعداد اعضای فهرست |
| جایگاه 6 | درجهٔ سه | `Θ(n³)` | ضرب معمولی دو ماتریس مربعی | تعداد سطر یا ستون هر ماتریس |
| جایگاه 7 | نمایی | `Θ(2ⁿ)` | بررسی همهٔ زیرمجموعه‌ها برای مجموع هدف | تعداد اعضای فهرست |
| جایگاه 8 | فاکتوریلی | `Θ(n!)` | پیمایش همهٔ ترتیب‌های کارهای متمایز | تعداد کارهای متمایز |


## آماده‌سازی رسم و شمارش

این سلول ابزار رسم `Matplotlib`، محاسبهٔ عددی `NumPy` و ابزار نمایش جدول‌ها را آماده می‌کند. مقدار تابع‌ها مستقیماً از روی فرمول محاسبه می‌شود. شکل‌های نوت‌بوک در پوشهٔ `figures` نیز ذخیره می‌شوند.


````python
from itertools import combinations, combinations_with_replacement, product
from math import factorial, log2
from pathlib import Path
import random
import numpy as np
import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display
from IPython.display import display, Markdown, HTML

# جهت متن آموزشی فارسی؛ فرمول‌ها و کدها چپ‌به‌راست می‌مانند.
display(HTML("""<style>
.jp-MarkdownOutput {direction: rtl; text-align: right;}
.jp-MarkdownOutput pre {direction: ltr; text-align: left;}
.jp-MarkdownOutput code {direction: ltr; unicode-bidi: isolate;}
.jp-MarkdownOutput table {direction: rtl;}
.jp-MarkdownOutput img {display: block; margin: 1em auto;}
</style>"""))

FIGURES = Path("figures")
FIGURES.mkdir(exist_ok=True)
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.22})

PERSIAN = ["ثابت", "لگاریتمی", "خطی", "خطی‌لگاریتمی", "درجهٔ دو", "درجهٔ سه", "نمایی", "فاکتوریلی"]
LABELS = [r"$1$", r"$\log_2 n$", r"$n$", r"$n\log_2 n$", r"$n^2$", r"$n^3$", r"$2^n$", r"$n!$"]
ORDERS = ["Θ(1)", "Θ(log n)", "Θ(n)", "Θ(n log n)", "Θ(n²)", "Θ(n³)", "Θ(2ⁿ)", "Θ(n!)"]
COLORS = ["#222222", "#0072b2", "#009e73", "#e69f00", "#cc79a7", "#56b4e9", "#d55e00", "#7b3294"]

def growth_values(n):
    return [1, log2(n), n, n * log2(n), n**2, n**3, 2**n, factorial(n)]

def persian_label(text):
    # اتصال حروف و ترتیب دیداری برای برچسب‌های فارسی در نمودار.
    return get_display(arabic_reshaper.reshape(text))

def show_table(headers, rows):
    lines = ["| " + " | ".join(headers) + " |", "|" + "---|" * len(headers)]
    lines += ["| " + " | ".join(str(value) for value in row) + " |" for row in rows]
    display(Markdown("\n".join(lines)))

def save_and_show(fig, filename):
    fig.savefig(FIGURES / filename, dpi=150, bbox_inches="tight", metadata={"Software": "Matplotlib"})
    plt.show()


def _tick(stats, name="steps"):
    """ثبت یک عملیات منتخب، در صورت درخواست شمارش."""
    if stats is not None:
        stats[name] = stats.get(name, 0) + 1
````

## نمودارهای رشد زمانی

روش رسم: برای هر مرتبه، تابع نماینده با ضریب یک رسم شده است. محور افقی اندازهٔ ورودی و محور عمودی مقدار تابع رشد است. این مقدار، واحد قراردادی کار است؛ نوت‌بوک همهٔ مقدارها و شکل‌ها را از روی فرمول‌ها محاسبه می‌کند.

### نمودار کلی هشت مرتبه

تصویر زیر از مقیاس لگاریتمی `Logarithmic Scale` در محور عمودی استفاده می‌کند تا هم تابع ثابت و هم تابع فاکتوریلی دیده شوند. فاصله‌های عمودی برابر در این مقیاس، نسبت‌های برابر را نشان می‌دهند. این مقیاس ترتیب مقدار تابع‌ها را تغییر نمی‌دهد.


### نمودار مرتبه‌های کم‌رشد و چندجمله‌ای

تصویر دوم دو نمای نزدیک‌تر دارد: سمت چپ ثابت تا خطی‌لگاریتمی و سمت راست خطی تا درجهٔ سه. در سمت چپ، خطی‌لگاریتمی به‌تدریج از خطی فاصله می‌گیرد. در سمت راست، فاصلهٔ درجهٔ سه و درجهٔ دو با افزایش اندازه بیشتر می‌شود.


### نمودار اثر دوبرابرشدن ورودی

تصویر سوم نسبت کار در اندازهٔ بیست به اندازهٔ ده را نشان می‌دهد. تابع ثابت تغییری نمی‌کند؛ تابع خطی دوبرابر، درجهٔ دو چهاربرابر و درجهٔ سه هشت‌برابر می‌شود. در تابع نمایی و فاکتوریلی، همین افزایش ورودی جهش بسیار بزرگ‌تری ایجاد می‌کند.


| مرتبه | نسبت کار در اندازهٔ بیست به ده |
|---|---|
| مرتبهٔ ثابت | 1.000 |
| مرتبهٔ لگاریتمی | 1.301 |
| مرتبهٔ خطی | 2.000 |
| مرتبهٔ خطی‌لگاریتمی | 2.602 |
| مرتبهٔ درجهٔ دو | 4.000 |
| مرتبهٔ درجهٔ سه | 8.000 |
| مرتبهٔ نمایی | 1,024.000 |
| مرتبهٔ فاکتوریلی | 670,442,572,800.000 |


````python
# نمودار نخست: هر هشت تابع روی یک بازهٔ یکسان.
n_small = list(range(2, 13))
fig, ax = plt.subplots(figsize=(10, 6))
for index, label in enumerate(LABELS):
    ax.plot(n_small, [growth_values(n)[index] for n in n_small],
            marker="o", markersize=4, color=COLORS[index], label=label)
ax.set(yscale="log", xlabel=persian_label("اندازهٔ ورودی"),
       ylabel=persian_label("مقدار تابع رشد؛ مقیاس لگاریتمی"),
       title=persian_label("مقایسهٔ رشد هشت مرتبهٔ زمانی"))
ax.set_xticks(n_small)
ax.legend(ncol=2, fontsize=11)
fig.tight_layout()
save_and_show(fig, "complexity_overview.png")

# نمودار دوم: دو نمای نزدیک‌تر از تابع‌های کم‌رشد و چندجمله‌ای.
n_large = list(range(2, 129))
fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))
for ax, indices, title in zip(axes, [(0, 1, 2, 3), (2, 3, 4, 5)],
                             ["ثابت تا خطی‌لگاریتمی", "خطی تا درجهٔ سه"]):
    for index in indices:
        ax.plot(n_large, [growth_values(n)[index] for n in n_large],
                color=COLORS[index], label=LABELS[index], linewidth=2)
    ax.set(yscale="log", xlabel=persian_label("اندازهٔ ورودی"),
           ylabel=persian_label("مقدار تابع رشد؛ مقیاس لگاریتمی"), title=persian_label(title))
    ax.legend(fontsize=11)
fig.tight_layout()
save_and_show(fig, "polynomial_growth.png")

# نمودار سوم: نسبت کار، هنگام دوبرابرشدن اندازه از ده به بیست.
base_n = 10
doubling_ratios = [new / old for new, old in zip(growth_values(2 * base_n), growth_values(base_n))]
fig, ax = plt.subplots(figsize=(10, 5.2))
bars = ax.bar(range(8), doubling_ratios, color=COLORS, width=0.68)
ax.set_xticks(range(8), LABELS, fontsize=12)
ax.set(yscale="log", ylabel=persian_label("نسبت کار در اندازهٔ بیست به ده"),
       title=persian_label("اثر دوبرابرشدن ورودی از ده به بیست"), ylim=(0.7, 1e14))
for bar, ratio in zip(bars, doubling_ratios):
    label = f"{ratio:.3g}" if ratio < 10000 else f"{ratio:.3e}"
    ax.text(bar.get_x() + bar.get_width() / 2, ratio * 1.3, label,
            ha="center", va="bottom", fontsize=10)
fig.tight_layout()
save_and_show(fig, "doubling_growth.png")
````

## مثال‌ها و فلوچارت‌ها

کد هر الگوریتم پس از توضیح و فلوچارت آمده است. شمارندهٔ اختیاری، عملیات اصلی همان مثال را ثبت می‌کند و پاسخ الگوریتم را تغییر نمی‌دهد. هر تصویر فلوچارت، پیوست داخل همین نوت‌بوک است.


### مرتبهٔ ثابت

معادل انگلیسی این مرتبه، `Constant Time` است. مثال الگوریتم، دسترسی مستقیم به عضو اول با مرتبهٔ `Θ(1)` است.

ایدهٔ الگوریتم: برای خواندن عضو اول یک فهرست غیرتهی، به اندیس صفر مراجعه می‌کنیم. بزرگ‌شدن فهرست تعداد این دسترسی‌ها را بیشتر نمی‌کند.

مثال اجرا:

```python
first_item([7, 2, 9, 4])
```

پاسخ مثال:

```text
7
```

دلیل مرتبه: یک بررسی خالی‌بودن و یک دسترسی مستقیم داریم؛ تعداد گام‌ها با اندازهٔ ورودی رشد نمی‌کند.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
T(n) = c = Θ(1)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([شروع]) --> B[دریافت فهرست]
    B --> C{فهرست خالی است؟}
    C -- بله --> E([اعلام خطای ورودی خالی])
    C -- خیر --> D[خواندن عضو با اندیس صفر]
    D --> F([بازگرداندن عضو اول])
```

</details>


````python
def first_item(values, stats=None):
    """دسترسی مستقیم به عضو اول؛ زمان ثابت برای فهرست غیرتهی."""
    if not values:
        raise ValueError("فهرست نباید خالی باشد.")
    _tick(stats)
    return values[0]

stats = {}
answer = first_item([7, 2, 9, 4], stats)
assert answer == 7 and stats['steps'] == 1
print('پاسخ:', answer, '؛ تعداد خواندن عضو:', stats['steps'])
````

### مرتبهٔ لگاریتمی

معادل انگلیسی این مرتبه، `Logarithmic Time` است. مثال الگوریتم، جست‌وجوی دودویی با مرتبهٔ `Θ(log n)` است.

ایدهٔ الگوریتم: جست‌وجوی دودویی `Binary Search` در هر گام عضو وسط را با هدف مقایسه می‌کند و یکی از دو نیمه را کنار می‌گذارد. داده‌ها باید از قبل صعودی مرتب باشند.

مثال اجرا:

```python
binary_search([2, 5, 8, 12, 16, 23, 38], 23)
```

پاسخ مثال:

```text
5
```

دلیل مرتبه: پس از هر گام، حدود نصف نامزدها می‌مانند. برای رسیدن از اندازهٔ ورودی به یک عضو، حدود لگاریتمِ اندازه گام لازم است. این مرتبه برای بدترین حالت است؛ پیدا شدن هدف در اولین میانه، زمان ثابت دارد. در حضور تکرار، این نسخه یکی از رخدادها را برمی‌گرداند.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
n → n/2 → n/4 → ... → 1
T(n) = T(floor(n/2)) + Θ(1) = Θ(log n)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([شروع با فهرست مرتب و هدف]) --> B["low = 0؛ high = n - 1"]
    B --> C{"low <= high؟"}
    C -- خیر --> Z([بازگرداندن منفی یک])
    C -- بله --> D["mid = low + (high - low) // 2"]
    D --> E{"values[mid] == target؟"}
    E -- بله --> F([بازگرداندن mid])
    E -- خیر --> G{"values[mid] < target؟"}
    G -- بله --> H["low = mid + 1"]
    G -- خیر --> I["high = mid - 1"]
    H --> C
    I --> C
```

</details>


````python
def binary_search(values, target, stats=None):
    """جست‌وجوی دودویی در فهرست صعودی؛ یک رخداد یا منفی یک."""
    low, high = 0, len(values) - 1
    while low <= high:
        _tick(stats)
        mid = low + (high - low) // 2
        if values[mid] == target:
            return mid
        if values[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

stats = {}
values = [2, 5, 8, 12, 16, 23, 38]
answer = binary_search(values, 23, stats)
assert answer == 5 and stats['steps'] == 2
print('اندیس پاسخ:', answer, '؛ تعداد گام:', stats['steps'])
````

### مرتبهٔ خطی

معادل انگلیسی این مرتبه، `Linear Time` است. مثال الگوریتم، یافتن بیشینه با پیمایش با مرتبهٔ `Θ(n)` است.

ایدهٔ الگوریتم: عضو اول را بیشینهٔ فعلی می‌گیریم؛ سپس هر عضو بعدی را یک‌بار با آن مقایسه و در صورت نیاز بیشینه را به‌روز می‌کنیم.

مثال اجرا:

```python
linear_max([7, 2, 9, 4])
```

پاسخ مثال:

```text
9
```

دلیل مرتبه: برای یک فهرست غیرتهی، تعداد مقایسه‌ها دقیقاً یک واحد کمتر از تعداد اعضاست. دوبرابرشدن تعداد اعضا، کار اصلی را تقریباً دوبرابر می‌کند.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
comparisons = n - 1
T(n) = Θ(n)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([شروع]) --> B{فهرست خالی است؟}
    B -- بله --> Z([اعلام خطای ورودی خالی])
    B -- خیر --> C["best = values[0]؛ index = 1"]
    C --> D{"index < n؟"}
    D -- خیر --> E([بازگرداندن best])
    D -- بله --> F{"values[index] > best؟"}
    F -- بله --> G["best = values[index]"]
    F -- خیر --> H["index = index + 1"]
    G --> H
    H --> D
```

</details>


````python
def linear_max(values, stats=None):
    """یافتن بیشینه با یک پیمایش؛ هر عضو بعدی یک مقایسه دارد."""
    if not values:
        raise ValueError("فهرست نباید خالی باشد.")
    best = values[0]
    for index in range(1, len(values)):
        _tick(stats)
        if values[index] > best:
            best = values[index]
    return best

stats = {}
answer = linear_max([7, 2, 9, 4], stats)
assert answer == 9 and stats['steps'] == 3
print('بیشینه:', answer, '؛ تعداد مقایسه:', stats['steps'])
````

### مرتبهٔ خطی‌لگاریتمی

معادل انگلیسی این مرتبه، `Linearithmic Time` است. مثال الگوریتم، مرتب‌سازی ادغامی با مرتبهٔ `Θ(n log n)` است.

ایدهٔ الگوریتم: مرتب‌سازی ادغامی `Merge Sort` فهرست را به دو نیمه تقسیم می‌کند، هر نیمه را مرتب می‌کند و نتیجه‌ها را ادغام می‌کند. این روش نمونهٔ تقسیم و حل `Divide and Conquer` است.

مثال اجرا:

```python
merge_sort([7, 2, 9, 4])
```

پاسخ مثال:

```text
[2, 4, 7, 9]
```

دلیل مرتبه: در هر سطحِ تقسیم، مجموع کار ادغام خطی است و حدود لگاریتمِ اندازه سطح داریم. حاصل، رشد خطی‌لگاریتمی است. تابع اصلی یک کپی و یک بافر می‌سازد؛ تابع `sort_range` بازه‌های نیمه‌باز را می‌گیرد و تابع `merge` دو نیمه را در بافر ادغام می‌کند. این نسخه همیشه ادغام را انجام می‌دهد، پس بهترین و بدترین زمان آن از همین مرتبه‌اند.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
T(n) = 2T(n/2) + Θ(n) = Θ(n log n)
For n = 2^k: buffer writes = n log2(n)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([ورود به تابع مرتب‌سازی بازه با دو مرز]) --> B{"hi - lo <= 1"}
    B -- بله --> Z([بازگشت])
    B -- خیر --> C["mid = (lo + hi) // 2"]
    C --> D[فراخوانی مرتب‌سازی بازه روی نیمهٔ چپ]
    D --> E[فراخوانی مرتب‌سازی بازه روی نیمهٔ راست]
    E --> F[فراخوانی تابع ادغام برای دو نیمه]
    F --> Z
```

</details>

توضیح تابع کمکی: در هر گام ادغام، کوچک‌ترین عضو ابتدای دو نیمه انتخاب می‌شود؛ سپس اعضای باقی‌مانده منتقل می‌شوند. انتخاب نیمهٔ چپ در حالت برابری، ترتیب اعضای برابر را حفظ می‌کند.



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([ورود به تابع ادغام]) --> B["left = lo; right = mid; out = lo"]
    B --> C{هر دو نیمه عضو بررسی‌نشده دارند؟}
    C -- بله --> D{"items[left] <= items[right]"}
    D -- بله --> E[انتقال عضو چپ به بافر و افزایش شاخص چپ]
    D -- خیر --> F[انتقال عضو راست به بافر و افزایش شاخص راست]
    E --> G["out = out + 1"]
    F --> G
    G --> C
    C -- خیر --> H[انتقال باقی‌مانده هر نیمه به بافر]
    H --> I[کپی بازهٔ بافر به همان بازه از فهرست]
    I --> Z([بازگشت])
```

</details>


````python
def merge_sort(values, stats=None):
    """مرتب‌سازی ادغامی؛ شمارنده تعداد نوشتن در بافر را ثبت می‌کند."""
    items = list(values)
    buffer = [None] * len(items)

    def merge(lo, mid, hi):
        left, right, out = lo, mid, lo
        while left < mid and right < hi:
            if items[left] <= items[right]:
                buffer[out] = items[left]
                left += 1
            else:
                buffer[out] = items[right]
                right += 1
            _tick(stats)
            out += 1
        while left < mid:
            buffer[out] = items[left]
            _tick(stats)
            left += 1
            out += 1
        while right < hi:
            buffer[out] = items[right]
            _tick(stats)
            right += 1
            out += 1
        for index in range(lo, hi):
            items[index] = buffer[index]

    def sort_range(lo, hi):
        if hi - lo <= 1:
            return
        mid = (lo + hi) // 2
        sort_range(lo, mid)
        sort_range(mid, hi)
        merge(lo, mid, hi)

    sort_range(0, len(items))
    return items

stats = {}
answer = merge_sort([7, 2, 9, 4], stats)
assert answer == [2, 4, 7, 9] and stats['steps'] == 8
print('فهرست مرتب:', answer, '؛ تعداد نوشتن در بافر:', stats['steps'])
````

### مرتبهٔ درجهٔ دو

معادل انگلیسی این مرتبه، `Quadratic Time` است. مثال الگوریتم، مرتب‌سازی انتخابی با مرتبهٔ `Θ(n²)` است.

ایدهٔ الگوریتم: مرتب‌سازی انتخابی `Selection Sort` در هر گذر کوچک‌ترین عضو بخش باقی‌مانده را پیدا می‌کند و در ابتدای آن بخش قرار می‌دهد.

مثال اجرا:

```python
selection_sort([7, 2, 9, 4])
```

پاسخ مثال:

```text
[2, 4, 7, 9]
```

دلیل مرتبه: گذر اول یک واحد کمتر از تعداد اعضا مقایسه دارد، گذر دوم دو واحد کمتر و همین‌طور تا یک مقایسه. جمع این اعداد درجهٔ دو است؛ حتی ورودی مرتب هم همین تعداد مقایسه دارد.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
comparisons = (n-1) + (n-2) + ... + 1 = n(n-1)/2
T(n) = Θ(n²)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([شروع]) --> B[کپی فهرست و قراردادن شاخص شروع روی صفر]
    B --> C{"first < n - 1؟"}
    C -- خیر --> Z([بازگرداندن فهرست مرتب])
    C -- بله --> D["minimum = first؛ index = first + 1"]
    D --> E{"index < n؟"}
    E -- بله --> F{"items[index] < items[minimum]؟"}
    F -- بله --> G["minimum = index"]
    F -- خیر --> H["index = index + 1"]
    G --> H
    H --> E
    E -- خیر --> I[جابجایی عضو کمینه با عضو ابتدای بخش در صورت نیاز]
    I --> J["first = first + 1"]
    J --> C
```

</details>


````python
def selection_sort(values, stats=None):
    """مرتب‌سازی انتخابی؛ خروجی جدید می‌سازد و ورودی را تغییر نمی‌دهد."""
    items = list(values)
    for first in range(len(items) - 1):
        minimum = first
        for index in range(first + 1, len(items)):
            _tick(stats)
            if items[index] < items[minimum]:
                minimum = index
        if minimum != first:
            items[first], items[minimum] = items[minimum], items[first]
    return items

stats = {}
answer = selection_sort([7, 2, 9, 4], stats)
assert answer == [2, 4, 7, 9] and stats['steps'] == 6
print('فهرست مرتب:', answer, '؛ تعداد مقایسه:', stats['steps'])
````

### مرتبهٔ درجهٔ سه

معادل انگلیسی این مرتبه، `Cubic Time` است. مثال الگوریتم، ضرب معمولی دو ماتریس مربعی با مرتبهٔ `Θ(n³)` است.

ایدهٔ الگوریتم: برای هر سطر ماتریس اول و هر ستون ماتریس دوم، حاصل‌ضرب درایه‌های متناظر را جمع می‌کنیم. سه حلقه روی سطر، ستون و اندیس میانی داریم.

مثال اجرا:

```python
matrix_multiply([[1, 2], [3, 4]], [[5, 6], [7, 8]])
```

پاسخ مثال:

```text
[[19, 22], [43, 50]]
```

دلیل مرتبه: سه حلقه هرکدام به اندازهٔ بعد ماتریس تکرار می‌شوند؛ تعداد ضرب‌های عددی دقیقاً مکعبِ آن بعد است. در این مثال، اندازهٔ ورودیِ نمودار تعداد سطر یا ستون هر ماتریس است و هر ماتریس مربعِ این تعداد درایه دارد.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
C[i][j] = sum(A[i][k] * B[k][j], k = 0,...,n-1)
scalar multiplications = n × n × n = n³
T(n) = Θ(n³)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([شروع]) --> B{دو ماتریس مربعی و هم‌اندازه‌اند؟}
    B -- خیر --> X([اعلام خطای ابعاد])
    B -- بله --> C[ساخت ماتریس نتیجه با درایه‌های صفر و آغاز از سطر صفر]
    C --> D{"i < n؟"}
    D -- خیر --> Z([بازگرداندن ماتریس نتیجه])
    D -- بله --> E["j = 0"]
    E --> F{"j < n؟"}
    F -- خیر --> G["i = i + 1"]
    G --> D
    F -- بله --> H["k = 0"]
    H --> I{"k < n؟"}
    I -- بله --> J["result[i][j] += left[i][k] * right[k][j]"]
    J --> K["k = k + 1"]
    K --> I
    I -- خیر --> L["j = j + 1"]
    L --> F
```

</details>


````python
def matrix_multiply(left, right, stats=None):
    """ضرب معمولی دو ماتریس مربعی؛ شمارنده تعداد ضرب‌های عددی است."""
    n = len(left)
    if len(right) != n or any(len(row) != n for row in left + right):
        raise ValueError("دو ماتریس باید مربعی و هم‌اندازه باشند.")
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                _tick(stats)
                result[i][j] += left[i][k] * right[k][j]
    return result

stats = {}
answer = matrix_multiply([[1, 2], [3, 4]], [[5, 6], [7, 8]], stats)
assert answer == [[19, 22], [43, 50]] and stats['steps'] == 8
print('ماتریس حاصل:', answer, '؛ تعداد ضرب:', stats['steps'])
````

### مرتبهٔ نمایی

معادل انگلیسی این مرتبه، `Exponential Time` است. مثال الگوریتم، بررسی همهٔ زیرمجموعه‌ها برای مجموع هدف با مرتبهٔ `Θ(2ⁿ)` است.

ایدهٔ الگوریتم: برای هر عضو دو تصمیم داریم: آن را انتخاب کنیم یا کنار بگذاریم. تابع بازگشتی هر دو تصمیم را بررسی می‌کند و در پایان می‌سنجد مجموع اعضای انتخاب‌شده برابر هدف هست یا نه.

مثال اجرا:

```python
count_subset_sums([1, 2, 3], 3)
```

پاسخ مثال:

```text
2
```

دلیل مرتبه: درخت انتخاب در هر سطح دو شاخه دارد؛ در پایان، دو به توان تعداد اعضا برگ و یک واحد کمتر از دو برابرِ آن گره داریم. در کد، مجموع جاری همراه فراخوانی منتقل می‌شود و در برگ، ساختن یا کپی‌کردن زیرمجموعه انجام نمی‌شود. با این کار، هزینهٔ هر گره ثابت می‌ماند. رخدادهای تکراری بر اساس موقعیتشان انتخاب می‌شوند.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
T(n) = 2T(n-1) + Θ(1) = Θ(2ⁿ)
leaves = 2ⁿ
nodes = 2^(n+1) - 1
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([ورود به تابع بازدید با شاخص عضو و مجموع جاری]) --> B{"index == n"}
    B -- بله --> C{مجموع برابر هدف است؟}
    C -- بله --> D([بازگرداندن یک])
    C -- خیر --> E([بازگرداندن صفر])
    B -- خیر --> F[بازدید عضو بعدی بدون افزودن عضو فعلی به مجموع]
    F --> G[بازدید عضو بعدی با افزودن عضو فعلی به مجموع]
    G --> H([بازگرداندن مجموع دو پاسخ])
```

</details>

توضیح پاسخ مثال: دو انتخاب موفق، عضو سوم به‌تنهایی و دو عضو اول با هم هستند.


````python
def count_subset_sums(values, target, stats=None):
    """شمارش زیرمجموعه‌های دارای مجموع هدف با انتخاب یا حذف هر عضو."""
    n = len(values)

    def visit(index, total):
        _tick(stats, "nodes")
        if index == n:
            _tick(stats, "leaves")
            return int(total == target)
        without_current = visit(index + 1, total)
        with_current = visit(index + 1, total + values[index])
        return without_current + with_current

    return visit(0, 0)

stats = {}
answer = count_subset_sums([1, 2, 3], 3, stats)
assert answer == 2 and stats['leaves'] == 8 and stats['nodes'] == 15
print('تعداد زیرمجموعه‌های با مجموع سه:', answer)
print('تعداد برگ‌ها:', stats['leaves'], '؛ تعداد کل گره‌ها:', stats['nodes'])
````

### مرتبهٔ فاکتوریلی

معادل انگلیسی این مرتبه، `Factorial Time` است. مثال الگوریتم، پیمایش همهٔ ترتیب‌های کارهای متمایز با مرتبهٔ `Θ(n!)` است.

ایدهٔ الگوریتم: برای جایگاه اول هرکدام از کارها را انتخاب می‌کنیم؛ برای جایگاه دوم یکی از کارهای باقی‌مانده را می‌گذاریم و همین روند را ادامه می‌دهیم. با عقب‌گرد `Backtracking` جابجایی قبلی برگردانده می‌شود تا انتخاب بعدی بررسی شود.

مثال اجرا:

```python
count_permutations(['الف', 'ب', 'پ', 'ت'])
```

پاسخ مثال:

```text
24
```

دلیل مرتبه: تعداد ترتیب‌های کامل، حاصل‌ضرب تعداد انتخاب‌ها در جایگاه‌ها و برابر فاکتوریل است. این کد فقط تعداد ترتیب‌ها را می‌شمارد و در هر برگ عدد یک برمی‌گرداند؛ کپی‌کردن کل ترتیب در برگ انجام نمی‌شود. تعداد کل گره‌ها کمتر از سه برابر تعداد برگ‌هاست، پس مرتبهٔ زمان خودِ این نسخه فاکتوریلی است. اگر کارها برچسب‌های یکسان داشته باشند، کد همچنان ترتیب موقعیت‌ها را می‌شمارد.

رابطهٔ شمارش یا رابطهٔ بازگشتی `Recurrence Relation`:

```text
leaves = n × (n-1) × ... × 1 = n!
T(n) = nT(n-1) + Θ(n) = Θ(n!)
```



<details>
<summary>متن مرمید فلوچارت</summary>

```mermaid
flowchart TD
    A([ورود به تابع بازدید با شاخص جایگاه فعلی]) --> B{"first == n"}
    B -- بله --> Z([بازگرداندن یک])
    B -- خیر --> C["total = 0؛ index = first"]
    C --> D{"index < n؟"}
    D -- خیر --> E([بازگرداندن تعداد ترتیب‌ها])
    D -- بله --> F[جابجایی عضو جایگاه فعلی با عضو انتخاب‌شده]
    F --> G[بازدید جایگاه بعدی و افزودن پاسخ به تعداد ترتیب‌ها]
    G --> H[بازگرداندن جابجایی قبلی]
    H --> I["index = index + 1"]
    I --> D
```

</details>

توضیح تعداد گره‌ها: برای چهار کار، مجموع گره‌های سطوح مختلف برابر یک به‌علاوهٔ چهار به‌علاوهٔ دوازده به‌علاوهٔ بیست‌وچهار به‌علاوهٔ بیست‌وچهار است؛ حاصل، شصت‌وپنج می‌شود.


````python
def count_permutations(values, stats=None):
    """شمارش ترتیب‌های موقعیت‌ها با پیمایش همهٔ جایگشت‌ها؛ بدون کپی در برگ."""
    items = list(values)
    n = len(items)

    def visit(first):
        _tick(stats, "nodes")
        if first == n:
            _tick(stats, "leaves")
            return 1
        total = 0
        for index in range(first, n):
            items[first], items[index] = items[index], items[first]
            total += visit(first + 1)
            # بازگرداندن وضعیت برای بررسی انتخاب بعدی.
            items[first], items[index] = items[index], items[first]
        return total

    return visit(0)

stats = {}
answer = count_permutations(['الف', 'ب', 'پ', 'ت'], stats)
assert answer == 24 and stats['leaves'] == 24 and stats['nodes'] == 65
print('تعداد ترتیب‌های کامل:', answer, '؛ تعداد کل گره‌ها:', stats['nodes'])
````

## بررسی تعداد عملیات روی کد

واحدهای شمارش: دسترسی به عضو اول، گام جست‌وجوی دودویی، مقایسه با بیشینه، نوشتن در بافر ادغام، مقایسه برای انتخاب کمینه، ضرب دو درایه، بررسی یک زیرمجموعه و بررسی یک ترتیب کامل. برای جست‌وجو، هدف بزرگ‌تر از تمام اعضا انتخاب می‌شود.


````python
ALGORITHM_NAMES = ['دسترسی مستقیم به عضو اول', 'جست\u200cوجوی دودویی', 'یافتن بیشینه با پیمایش', 'مرتب\u200cسازی ادغامی', 'مرتب\u200cسازی انتخابی', 'ضرب معمولی دو ماتریس مربعی', 'بررسی همهٔ زیرمجموعه\u200cها برای مجموع هدف', 'پیمایش همهٔ ترتیب\u200cهای کارهای متمایز']


def observed_counts(n):
    values = list(range(n))
    counters = []
    for function in (first_item, binary_search, linear_max, merge_sort, selection_sort):
        stats = {}
        if function is binary_search:
            function(values, n, stats)  # هدف بزرگ‌تر از تمام اعضاست.
        else:
            function(values, stats)
        counters.append(stats.get("steps", 0))
    stats = {}
    matrix_multiply([[1] * n for _ in range(n)], [[1] * n for _ in range(n)], stats)
    counters.append(stats["steps"])
    stats = {}
    count_subset_sums(values, n, stats)
    counters.append(stats["leaves"])
    stats = {}
    count_permutations(values, stats)
    counters.append(stats["leaves"])
    return counters

actual_counts = {n: observed_counts(n) for n in (2, 4, 8)}
show_table(["الگوریتم", "اندازهٔ ۲", "اندازهٔ ۴", "اندازهٔ ۸"],
           [(name, *[f"{actual_counts[n][index]:,}" for n in (2, 4, 8)])
            for index, name in enumerate(ALGORITHM_NAMES)])

for n, counts in actual_counts.items():
    assert counts == [1, n.bit_length(), n - 1, n * int(log2(n)),
                      n * (n - 1) // 2, n**3, 2**n, factorial(n)]
print("نتیجه: شمار عملیات منتخب با فرمول‌های تحلیل‌شده تطبیق دارد.")
````

## بررسی درستی پاسخ‌ها

آزمون‌ها ورودی‌های خالی و تک‌عضوی، مقدارهای تکراری، هدف غایب، همهٔ فهرست‌های کوتاه، ضرب در ماتریس همانی و صفر، ابعاد نامعتبر، زیرمجموعه‌ها و تعداد جایگشت‌ها را بررسی می‌کنند. مرتب‌سازی مرجع آماده فقط در بررسی پاسخ به‌کار می‌رود؛ خود الگوریتم‌ها مستقل پیاده‌سازی شده‌اند.


````python
def run_checks():
    """بررسی پاسخ‌ها با مرجع مستقل و کنترل شمار عملیات منتخب."""
    totals = {"دسترسی و بیشینه": 0, "جست‌وجو": 0, "مرتب‌سازی": 0,
              "ضرب ماتریس": 0, "زیرمجموعه": 0, "جایگشت": 0, "شمار عملیات": 0}
    for function in (first_item, linear_max):
        try:
            function([])
        except ValueError:
            totals["دسترسی و بیشینه"] += 1
        else:
            raise AssertionError("ورودی خالی باید خطای مشخص داشته باشد.")

    for n in range(7):
        for values in product(range(3), repeat=n):
            values = list(values)
            before = values.copy()
            for function in (selection_sort, merge_sort):
                assert function(values) == sorted(values)
                assert values == before
                totals["مرتب‌سازی"] += 1
            if values:
                assert first_item(values) == values[0]
                assert linear_max(values) == max(values)
                totals["دسترسی و بیشینه"] += 2

    for n in range(7):
        for values in combinations_with_replacement(range(5), n):
            values = list(values)
            for target in range(-1, 6):
                answer = binary_search(values, target)
                if target in values:
                    assert 0 <= answer < len(values) and values[answer] == target
                else:
                    assert answer == -1
                totals["جست‌وجو"] += 1

    assert matrix_multiply([[1, 2], [3, 4]], [[5, 6], [7, 8]]) == [[19, 22], [43, 50]]
    totals["ضرب ماتریس"] += 1
    rng = random.Random(20261008)
    for n in range(6):
        for _ in range(10):
            matrix = [[rng.randrange(-5, 6) for _ in range(n)] for _ in range(n)]
            identity = [[int(i == j) for j in range(n)] for i in range(n)]
            zero = [[0] * n for _ in range(n)]
            assert matrix_multiply(matrix, identity) == matrix
            assert matrix_multiply(identity, matrix) == matrix
            assert matrix_multiply(matrix, zero) == zero
            totals["ضرب ماتریس"] += 3
    for left, right in [([[1, 2]], [[1]]), ([[1]], [[1, 2]]), ([], [[1]])]:
        try:
            matrix_multiply(left, right)
        except ValueError:
            totals["ضرب ماتریس"] += 1
        else:
            raise AssertionError("ابعاد نامعتبر باید تشخیص داده شوند.")

    for n in range(6):
        for values in product(range(3), repeat=n):
            for target in range(-1, 9):
                expected = sum(sum(chosen) == target for size in range(n + 1)
                               for chosen in combinations(values, size))
                assert count_subset_sums(values, target) == expected
                totals["زیرمجموعه"] += 1

    for n in range(9):
        data = list(range(n))
        before = data.copy()
        stats = {}
        assert count_permutations(data, stats) == factorial(n)
        assert stats["leaves"] == factorial(n)
        assert stats["nodes"] == sum(factorial(n) // factorial(n - k) for k in range(n + 1))
        assert data == before
        totals["جایگشت"] += 1

    for n in (1, 2, 4, 8, 16):
        data = list(range(n))
        stats = {}
        first_item(data, stats)
        assert stats["steps"] == 1
        stats = {}
        binary_search(data, n, stats)
        assert stats["steps"] == n.bit_length()
        stats = {}
        linear_max(data, stats)
        assert stats.get("steps", 0) == n - 1
        stats = {}
        merge_sort(data, stats)
        assert stats.get("steps", 0) == n * (n.bit_length() - 1)
        stats = {}
        selection_sort(data, stats)
        assert stats.get("steps", 0) == n * (n - 1) // 2
        totals["شمار عملیات"] += 5
    for n in range(7):
        stats = {}
        matrix_multiply([[1] * n for _ in range(n)], [[1] * n for _ in range(n)], stats)
        assert stats.get("steps", 0) == n**3
        stats = {}
        count_subset_sums(list(range(n)), 3, stats)
        assert stats["leaves"] == 2**n and stats["nodes"] == 2 ** (n + 1) - 1
        totals["شمار عملیات"] += 2
    return totals

for label, total in run_checks().items():
    print(f'بررسی {label}: {total} حالت موفق')
print('نتیجه: همهٔ بررسی‌ها موفق بودند.')
````
